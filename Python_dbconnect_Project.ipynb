```python
import mysql.connector
```

```python
#connect()=> method to connect the server
con=mysql.connector.connect(
    host="localhost",
    user="root",
    password="YOUR_MYSQL_PASSWORD"
)
#cursor()=> it is used to carry all operations on database
mycursor=con.cursor()

#execute()=>  function which executes the given queries.
#your queries should have;
mycursor.execute("create database if not exists db_demos")
#if not exists=> it will check if db/table already exists

con.close() #close connection for security purpose
```

```python
#Create table Person with p_name,age,salary
con=mysql.connector.connect(
    host="localhost",
    user="root",
    password="YOUR_MYSQL_PASSWORD",
    database='db_demos'
)

#cursor()=> it is used to carry all operations on database
mycursor=con.cursor()

#create a table person
sql='''create table if not exists Person
(p_name varchar(50) not null,age int,salary decimal(8,2))'''

mycursor.execute(sql)
print("Table Created Success!!")
con.close() #close connection for security purpose
```

```python
#Insert the Records in the Table
con=mysql.connector.connect(
    host="localhost",
    user="root",
    password="YOUR_MYSQL_PASSWORD",
    database='db_demos'
)

#cursor()=> it is used to carry all operations on database
mycursor=con.cursor()

#add single row
#%s is a placeholder for each value passed to database
mycursor.execute('''insert into Person(p_name,age,salary)values(%s,%s,%s)''',
('Peter',23,45000.78))

con.commit() #save to database
#rowcount - gives number of rows affected/added
print(mycursor.rowcount)

#add multiple rows
sqls="insert into Person(p_name,age,salary) values(%s,%s,%s)"

#use list of tuples having vlues for each row
vals=[('John',34,89000.87),('Janny',28,65400.8),('Olvia',20,32590.00)]

#executemany=> it will execute some statements multiple times
mycursor.executemany(sqls,vals)
con.commit() #save to database
print(mycursor.rowcount)

#returns id of last added row
print(mycursor.lastrowid)
con.close()
```

```python
#fetch all records from person table
import mysql.connector
con=mysql.connector.connect(
    host="localhost",
    user="root",
    password="YOUR_MYSQL_PASSWORD",
    database='db_demos'
)

#cursor()=> it is used to carry all operations on database
mycursor=con.cursor()

#fetch the records from person table
sql="select * from Person"

mycursor.execute(sql)

#fetch=>It will get all records which are the result of given query
#fetch=> It will get single/first record

result=mycursor.fetchall()
for rows in result:
    print(rows)

#columns can be accessed by using index number
print("Only Person Name & Age:")
for row in result:
    print(f'{row[0]} | {row[1]}')
```
